```python
import os
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
from matplotlib.lines import Line2D
import matplotlib as mpl

# 1. 设置学术排版风格 (保持与您之前脚本一致)
mpl.rcParams.update({
    "font.family": "sans-serif",
    "font.sans-serif": ["DejaVu Sans", "Arial", "Liberation Sans"],
    "font.size": 9,
    "axes.labelsize": 9,
    "axes.linewidth": 0.6,
    "figure.facecolor": "white",
    "savefig.facecolor": "white",
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
})

# 数据根目录
base_path = "/Users/jannik/Library/CloudStorage/OneDrive-个人/Study-Materials-and-Documents/Master/Masterarbeit/撰写/data/图4,2有问题的配准模态"

# 2. 精选元数据 (仅保留 A 和 D，并使用中性描述)
# 注意：Panel B 对应的位置现在由原 Panel D 占据，为了排版逻辑，我们标记为 Panel A 和 Panel B
images_metadata = [
    {
        "file": "YHC04_350.png",
        "title": "Panel A: Geometric Mismatch",
        "mode": "Geometric mismatch and residual artifacts",
    },
    {
        "file": "PDP04_351.png",
        "title": "Panel B: Mask Inconsistency", # 原 D 变更为 B，方便读者阅读
        "mode": "Masking inconsistency and local signal voids",
    }
]

def create_thesis_figure():
    # 改为 1x2 布局，调整 figsize 使比例更协调
    fig, axes = plt.subplots(1, 2, figsize=(8, 5), dpi=300)
    plt.subplots_adjust(wspace=0.15)
    
    for i, ax in enumerate(axes.flat):
        meta = images_metadata[i]
        img_path = os.path.join(base_path, meta["file"])
        
        try:
            img = mpimg.imread(img_path)
            ax.imshow(img, cmap='gray')
        except Exception:
            ax.text(0.5, 0.5, "Image not found", ha='center', va='center')
        
        ax.set_axis_off()
        
        # Panel 标题
        ax.text(0.5, 1.05, meta["title"], transform=ax.transAxes, 
                fontsize=10, fontweight='bold', ha='center')
        
        # 中性化的描述文字
        ax.text(0.5, -0.05, meta["mode"], transform=ax.transAxes, 
                fontsize=9, style='italic', ha='center', va='top')
        
        # 2cm 下界短横线与中心对准点
        line_y = -0.15
        ax.add_line(Line2D([0.4, 0.6], [line_y, line_y], transform=ax.transAxes, 
                           color='black', linewidth=1.0))
        ax.plot(0.5, line_y, marker='.', color='black', transform=ax.transAxes, markersize=4)

    # 保存为高质量文件
    output_base = "Figure_4_2_Selected_QC"
    fig.savefig(f"{output_base}.png", bbox_inches='tight', pad_inches=0.1)
    fig.savefig(f"{output_base}.pdf", bbox_inches='tight', pad_inches=0.1)
    plt.show()

if __name__ == "__main__":
    create_thesis_figure()
```
