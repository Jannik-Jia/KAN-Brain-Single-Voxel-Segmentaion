```python
import pandas as pd
import numpy as np

def compute_paper_metrics(csv_file_path):
    """
    读取交叉验证结果CSV文件，计算论文所需的 Mean 和 Std。
    """
    try:
        df = pd.read_csv(csv_file_path)
        print(f"成功加载文件: {csv_file_path}, 共 {len(df)} 个 Fold。")
    except FileNotFoundError:
        print(f"错误: 找不到文件 {csv_file_path}")
        return

    # 论文指标与CSV列名的映射关系
    # Key: 论文中使用的名称
    # Value: CSV中的列名
    metrics_map = {
        "Top-1 Accuracy": "test_gross_acc",
        "Balanced Accuracy": "test_balanced_acc",
        "Macro-F1": "test_macro_f1",
        "NLL": "test_nll",
        "ECE (Soft)": "test_soft_ece",
        "Brier Score": "test_brier",
        "3D Soft Dice": "test_3d_soft_dice_macro",
        "Risk (Soft) @95% Cov": "test_risk_soft_at_95cov",
        "Risk (Hard) @95% Cov": "test_risk_hard_at_95cov",
        "Coverage @5% Risk (Soft)": "test_coverage_at_5pct_risk_soft"
    }

    print("\n" + "="*60)
    print(f"{'Metric Name':<30} | {'Mean':<10} | {'Std':<10}")
    print("-" * 60)

    results = {}

    for paper_name, col_name in metrics_map.items():
        if col_name in df.columns:
            # 计算均值和样本标准差 (ddof=1)
            mean_val = df[col_name].mean()
            std_val = df[col_name].std()
            
            print(f"{paper_name:<30} | {mean_val:.4f}     | {std_val:.4f}")
            results[paper_name] = {'mean': mean_val, 'std': std_val}
        else:
            print(f"{paper_name:<30} | {'Not Found':<10} | {'N/A':<10}")
    
    print("="*60 + "\n")
    return results

# --- 使用示例 ---

# 1. 计算 Soft+CEST (你当前上传的文件)
print("计算 Soft+CEST 配置的数值:")
soft_metrics = compute_paper_metrics('/Users/jannik/Library/CloudStorage/OneDrive-个人/Study-Materials-and-Documents/Master/Masterarbeit/撰写/data/softlabel等数据/soft+cest/runs/leave_one_out_soft/cross_validation_results_soft.csv')

# 2. (可选) 如果你上传了 Hard+CEST 的文件，取消下面注释并修改文件名
# print("计算 Hard+CEST 配置的数值:")
hard_cest_metrics = compute_paper_metrics('/Users/jannik/Library/CloudStorage/OneDrive-个人/Study-Materials-and-Documents/Master/Masterarbeit/撰写/data/softlabel等数据/hard+cest/runs/leave_one_out_hard/cross_validation_results_hard.csv')

# 3. (可选) 如果你上传了 Hard+MPRAGE 的文件
# print("计算 Hard+MPRAGE 配置的数值:")
# hard_mprage_metrics = compute_paper_metrics('cross_validation_results_hard_mprage.csv')
```
