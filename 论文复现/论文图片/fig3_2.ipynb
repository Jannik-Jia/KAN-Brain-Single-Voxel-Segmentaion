```python
import matplotlib.pyplot as plt
import numpy as np

# ==========================================
# 1. 数据准备 (来自 Table 3.5)
# ==========================================
# Mean values
means_global = [58.5, 43.8]  # [GC, Macro-F1]
means_patient = [60.1, 46.5]

# Standard Deviations (用于误差棒)
stds_global = [9.5, 8.9]
stds_patient = [8.2, 7.8]

# 标签设置
metrics = ['Gross Accuracy (GC)', 'Macro-F1']
x = np.arange(len(metrics))  # X轴刻度位置
width = 0.35                 # 柱状图宽度

# ==========================================
# 2. 绘图设置 (符合学术出版标准)
# ==========================================
# 设置画布大小 (英寸) 和分辨率 (DPI)
fig, ax = plt.subplots(figsize=(6, 5), dpi=300)

# 定义配色: 浅色作为 Baseline, 深色作为 Ours
color_global = '#B0C4DE' # LightSteelBlue
color_patient = '#003366' # Dark Blue (Professional/Academic)

# 绘制柱状图 (带误差棒)
# capsize控制误差棒顶端横线的宽度
rects1 = ax.bar(x - width/2, means_global, width, yerr=stds_global, 
                label='Global StandardScaler', color=color_global, 
                capsize=5, edgecolor='black', linewidth=0.8, 
                error_kw=dict(lw=1, capthick=1))

rects2 = ax.bar(x + width/2, means_patient, width, yerr=stds_patient, 
                label='Per-Patient Z-Score', color=color_patient, 
                capsize=5, edgecolor='black', linewidth=0.8, 
                error_kw=dict(lw=1, capthick=1))

# ==========================================
# 3. 图表美化与标注
# ==========================================
ax.set_ylabel('Performance (%)', fontsize=12, fontweight='medium')
ax.set_xticks(x)
ax.set_xticklabels(metrics, fontsize=11)
ax.set_ylim(0, 80) # 留出空间给误差棒和图例

# 图例设置 (无边框或细边框，位于右上方)
ax.legend(loc='upper right', fontsize=10, frameon=True, 
          fancybox=False, edgecolor='black', framealpha=1.0)

# 添加背景网格 (仅Y轴)
ax.yaxis.grid(True, linestyle='--', which='major', color='grey', alpha=0.5)
ax.set_axisbelow(True) # 确保网格在柱子下方

# 去除顶部和右侧的边框 (Tufte style / Scientific style)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# 调整布局
plt.tight_layout()

# 保存
plt.show()
```
